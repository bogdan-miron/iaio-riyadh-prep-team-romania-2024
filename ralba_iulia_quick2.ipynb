```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
corr_matrix = df.corr()

# Plot correlation matrix using seaborn
plt.figure(figsize=(20, 18))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt="0.0001", linewidths=0.5)
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
plt.figure(figsize=(15, 10))
sns.boxplot(x='HATYEAR', y='VARSTA', data=df, palette='coolwarm')
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(df_scaled)

plt.figure(figsize=(15, 10))
sns.scatterplot(x=df['FEDFIELD'], y=df['VARSTA'], hue=df['Cluster'], palette='coolwarm', s=100)  # AICI MODIFICATI AXELE IN FUNCTIE DE CUM VA TRBEUI
plt.title("K-Means Clustering")
plt.xlabel("NFECOURSE")
plt.ylabel("VARSTA")
plt.legend(title="Cluster")
plt.show()
```
